```python
# Q.1 if you want to print largest number.

num = [20,40,60,90,10]
big = num[0]

for i in num:
  if (i > big):
    big = i
print(big)
```

```python
# Q.2 if you want to print smallest number.

num = [20,40,60,90,10]
small = num[0]

for i in num:
  if (i < small):
    small = i
print(small)
```

```python
# Q.3 if you want to fine sum of all numbers.

num = [10,5,15,20,30]
total = 0
for i in num:
  total= total + i

print(total)
```

```python
# Q.4 if you want to print only str variable.

student = [20,21,30,"kumar","bharat"]
for i in student:
  if(type(i) == str):
    print(i)
```
